### Decision 2: Missing Descriptions

**Affected records:** 1,454

**Decision:** Controlled recovery of missing descriptions.

Following the investigation, the affected StockCodes were classified into three groups:

- 674 StockCodes have exactly one distinct description elsewhere in the dataset.
- 174 StockCodes have multiple descriptions elsewhere.
- 112 StockCodes have no description elsewhere.

**Selected approach:**

Recover descriptions only where a reliable match can be established. Records with multiple possible descriptions or no matching description will remain missing.

**Justification:**

This approach allows us to recover potentially useful information without making unsupported assumptions. It also preserves the integrity of the original dataset.

**Implementation safeguards:**
- Keep the original dataset unchanged.
- Perform recovery in a separate working copy.
- Verify matching descriptions before filling.
- Document all recovered values.
- Leave uncertain descriptions missing.

### Decision 3: Negative UnitPrice and Bad Debt Adjustments

**Affected records:** 2 negative-price records and 1 related positive-price record.

**Decision:** Retain and flag separately.

The investigation identified two negative UnitPrice records, both described as "Adjust bad debt", with a UnitPrice of -£11,062.06 each. A related positive-price record of £11,062.06 was also identified.

All three records will be retained in the dataset but treated as special accounting adjustments rather than ordinary product sales.

**Justification:**
Removing these records could result in the loss of relevant accounting information. Separating them allows ordinary product sales analysis to remain meaningful while preserving the original entries.

**Implementation:**
- Retain all three records.
- Flag them as special adjustments.
- Exclude them from ordinary product sales calculations where appropriate.
- Preserve the original dataset unchanged.

### Decision 4: Cancellation and Return Records

**Affected records:** 10,624

**Decision:** Retain and classify separately.

The investigation identified:

- 9,288 cancellation records with negative quantities and positive unit prices.
- 1,336 negative-quantity records with zero unit prices, provisionally classified as operational or inventory adjustments.

**Justification:**

These records contain useful information about cancellations and inventory-related activities. Removing them would result in information loss and could distort subsequent analysis.

**Implementation:**
- Retain all records.
- Flag cancellation invoices separately.
- Flag negative-quantity, zero-price records as operational adjustments.
- Exclude these categories from ordinary sales calculations where appropriate.
- Preserve the original transaction values and quantities.

### Decision 5: Outlier Treatment

**Affected fields:** Quantity, UnitPrice and TransactionValue.

**Decision:** Retain original values and flag unusual observations where necessary.

The investigation identified:

- 56,363 Quantity outlier records.
- 37,999 UnitPrice outlier records.
- 42,651 TransactionValue outlier records.

**Justification:**

The IQR method identifies statistically unusual values but does not establish that they are incorrect. Further investigation revealed legitimate bulk purchases, special charges, accounting adjustments and unusually large transactions.

**Implementation:**
- Retain all original values.
- Do not delete or cap outliers automatically.
- Flag unusual observations for further investigation.
- Treat special charges and accounting adjustments separately where appropriate.
- Consider the influence of extreme values when interpreting analytical results.

### Decision 6: Exact Duplicates and Repeated Invoice Lines

**Decision:** Retain all records and flag exact duplicates.

The investigation identified 5,268 additional exact duplicate occurrences and 9,694 repeated invoice-product combinations.

Repeated invoice-product combinations were further classified into completely identical, mixed and different-row groups.

**Justification:**

Repeated records are not automatically errors. Some may represent legitimate transaction entries, while others may require further investigation.

Automatically deleting them could result in the loss of valid information.

**Implementation:**
- Retain all original records.
- Do not automatically remove exact duplicates.
- Flag exact duplicate occurrences in the working dataset.
- Preserve repeated invoice-product combinations with different transaction details.
- Consider the findings when interpreting transaction and product analysis.
