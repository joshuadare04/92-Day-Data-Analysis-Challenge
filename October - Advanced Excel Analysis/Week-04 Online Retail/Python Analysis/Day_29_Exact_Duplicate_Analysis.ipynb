```python
import pandas as pd
```

```python
file_path = r"C:\Users\JAY\Desktop\92 Day Data Analysis Challenge\October - Advanced Excel Analysis\Week-04 Online Retail\Source data\Online Retail.xlsx"

df = pd.read_excel(file_path)

print("Dataset loaded successfully!")
print("Rows:", len(df))
print("Columns:", len(df.columns))
```

```python
duplicate_rows = df[df.duplicated(keep=False)]
duplicate_patterns = duplicate_rows.drop_duplicates()

print("Total records in dataset:", len(df))
print("Exact duplicate occurrences:", df.duplicated().sum())
print("Rows involved in duplicate groups:", len(duplicate_rows))
print("Distinct duplicated row patterns:", len(duplicate_patterns))
```

```python
duplicate_rows.head(10)
```

```python
duplicate_frequency = (
    df[df.duplicated(keep=False)]
    .value_counts()
    .reset_index(name="Occurrences")
    .sort_values("Occurrences", ascending=False)
)

duplicate_frequency.head(10)
```

```python
row_frequencies = df.value_counts(dropna=False)

duplicate_frequencies = row_frequencies[row_frequencies > 1]

frequency_distribution = (
    duplicate_frequencies
    .value_counts()
    .sort_index()
    .rename_axis("Times_Appear")
    .reset_index(name="Distinct_Row_Patterns")
)

frequency_distribution
```

```python
duplicate_invoice_counts = (
    duplicate_rows.groupby("InvoiceNo")
    .size()
    .sort_values(ascending=False)
    .reset_index(name="Rows_In_Duplicate_Groups")
)

duplicate_invoice_counts.head(10)
```

```python
invoice_555524 = df[df["InvoiceNo"] == 555524]

print("Total records in invoice:", len(invoice_555524))
print("Distinct complete row patterns:", len(invoice_555524.drop_duplicates()))
print("Exact duplicate occurrences:", invoice_555524.duplicated().sum())
```

```python
invoice_duplicate_patterns = (
    invoice_555524
    .value_counts()
    .loc[lambda counts: counts > 1]
    .reset_index(name="Occurrences")
    .sort_values("Occurrences", ascending=False)
)

invoice_duplicate_patterns
```

```python
duplicate_pattern_invoice_counts = (
    duplicate_rows.groupby(["InvoiceNo", "StockCode", "Description"])
    .size()
    .reset_index(name="Rows_In_Duplicate_Groups")
    .sort_values("Rows_In_Duplicate_Groups", ascending=False)
)

duplicate_pattern_invoice_counts.head(10)
```

```python
# Identify extra occurrences of exact duplicate records
duplicate_occurrences = df[df.duplicated(keep="first")]

# Count extra duplicate occurrences by invoice
duplicate_occurrences_by_invoice = (
    duplicate_occurrences.groupby("InvoiceNo")
    .size()
    .sort_values(ascending=False)
    .reset_index(name="Exact_Duplicate_Occurrences")
)

print("Total exact duplicate occurrences:", len(duplicate_occurrences))

print(
    "Invoices containing exact duplicate occurrences:",
    duplicate_occurrences_by_invoice["InvoiceNo"].nunique()
)

print("\nTop 10 invoices by exact duplicate occurrences:")

duplicate_occurrences_by_invoice.head(10)
```
