```python
duplicate_analysis = (
    df.groupby(["InvoiceNo", "StockCode"])
    .apply(lambda group: pd.Series({
        "Total_Records": len(group),
        "Distinct_Complete_Rows": len(group.drop_duplicates())
    }))
    .reset_index()
)

duplicate_analysis = duplicate_analysis[
    duplicate_analysis["Total_Records"] > 1
]

duplicate_analysis.head()
```

```python
print("Number of repeated invoice-product combinations:",
      len(duplicate_analysis))

duplicate_analysis.head(10)
```

```python
duplicate_analysis["Exact_Duplicate_Occurrences"] = (
    duplicate_analysis["Total_Records"] -
    duplicate_analysis["Distinct_Complete_Rows"]
)

duplicate_analysis["Contains_Exact_Duplicates"] = (
    duplicate_analysis["Exact_Duplicate_Occurrences"] > 0
)

duplicate_analysis["Contains_Different_Rows"] = (
    duplicate_analysis["Distinct_Complete_Rows"] > 1
)

duplicate_analysis.head(10)
```

```python
total_combinations = len(duplicate_analysis)

contains_exact = duplicate_analysis[
    duplicate_analysis["Contains_Exact_Duplicates"]
]

contains_different = duplicate_analysis[
    duplicate_analysis["Contains_Different_Rows"]
]

completely_identical = duplicate_analysis[
    duplicate_analysis["Distinct_Complete_Rows"] == 1
]

print("Total repeated combinations:", total_combinations)

print("Combinations containing exact duplicates:",
      len(contains_exact))

print("Combinations containing different rows:",
      len(contains_different))

print("Completely identical combinations:",
      len(completely_identical))

print("Total additional exact duplicate occurrences:",
      duplicate_analysis["Exact_Duplicate_Occurrences"].sum())
```

```python
mixed_combinations = duplicate_analysis[
    (duplicate_analysis["Contains_Exact_Duplicates"]) &
    (duplicate_analysis["Contains_Different_Rows"])
]

different_only = duplicate_analysis[
    (~duplicate_analysis["Contains_Exact_Duplicates"]) &
    (duplicate_analysis["Contains_Different_Rows"])
]

print("Completely identical:", len(completely_identical))
print("Mixed combinations:", len(mixed_combinations))
print("Different rows only:", len(different_only))

print("Total:",
      len(completely_identical) +
      len(mixed_combinations) +
      len(different_only))
```
